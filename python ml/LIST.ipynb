```python
mylist=['aditya',10,'tomar',59,'AMITY']
mylist.insert(2,'university')
print(mylist)
mylist.append('aump')
print(mylist)
```

merge two list by using extend

```python
mylist=['aditya',10,'tomar',59,'AMITY']
mylist1=['aditya',67]
mylist.extend(mylist1)
print(mylist)
```

we can remove the element too with remove () method 

```python
mylist=['aditya',10,'tomar',59,'AMITY']
mylist.remove(10)
print(mylist)
```

```python
mylist=['aditya',10,'tomar',59,'AMITY']
mylist.pop(2)
print(mylist)
del mylist[2]
print(mylist)
```

```python
mylist=['aditya',10,'tomar',59,'AMITY']
del mylist[-1]
print(mylist)
del mylist[:2]
print(mylist)
```
