# Experiment 7 — Missing Values Techniques on Titanic Dataset
**Source:** Titanic PDF

The notebook implements the missing-value techniques shown in the PDF: row deletion, column deletion, arbitrary-value replacement, mean, median, max, min, forward fill, backward fill, and interpolation.

```python
import pandas as pd
import numpy as np

df=pd.read_csv("titanic_Data_Train.csv")
df.head()
```

```python
df.shape
df.tail()
```

```python
df.isnull().sum()
```

## 1. Deleting the Missing Values — Entire Row

```python
# axis=0 means rows; axis=1 means columns
df1=df.dropna(axis=0)
print(df1.shape)
print(df1.isnull().sum())
```

## 2. Deleting the Missing Values — Entire Column

```python
df2=df.drop(['Age','Cabin','Embarked'],axis=1)
print(df2.isnull().sum())
```

## 3. Imputing Missing Values — Arbitrary Value

```python
df3=df.copy()
df3['Age']=df3['Age'].fillna(0)
df3['Cabin']=df3['Cabin'].fillna(0)
df3['Embarked']=df3['Embarked'].fillna(0)
print(df3.isnull().sum())
```

## 4. Imputing Missing Values — Mean

```python
df4=df.copy()
df4['Age']=df4['Age'].fillna(df4['Age'].mean())
print(df4.isnull().sum())
```

## 5. Imputing Missing Values — Median

```python
df6=df.copy()
df6['Age']=df6['Age'].fillna(df6['Age'].median())
print(df6.isnull().sum())
```

## 6. Imputing Missing Values — Maximum

```python
df7=df.copy()
df7['Age']=df7['Age'].fillna(df7['Age'].max())
print(df7.isnull().sum())
```

## 7. Imputing Missing Values — Minimum

```python
df8=df.copy()
df8['Age']=df8['Age'].fillna(df8['Age'].min())
print(df8.isnull().sum())
```

## 8. Forward Fill

```python
df9=df.copy()
df9['Cabin'] = df9['Cabin'].ffill()
print(df9.isnull().sum())
```

## 9. Backward Fill

```python
df10=df.copy()
df10['Cabin'] = df10['Cabin'].bfill()
print(df10.isnull().sum())
```

## 10. Interpolation

```python
df11=df.copy()
df11['Age'] = df11['Age'].interpolate()
print(df11.isnull().sum())
```
