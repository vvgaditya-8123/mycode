# Experiment 5 — Linear Regression on Salary Data
**Source:** Linear Regression PDF

Implementation of Linear Regression on Salary Data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Salary_Data.csv')
df.head(10)
```

```python
df.tail()
df.shape
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,1]

print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,shuffle=True
)

print(x_train)
print(y_train)
print(x_test)
print(y_test)
```

```python
from sklearn.linear_model import LinearRegression

model=LinearRegression()
model.fit(x_train,y_train) # fit() function used for training a model
model.score(x_test,y_test) # score() function used for finding accuracy
```

```python
model.coef_ # value of 'm'
model.intercept_ # value of 'c'
```

```python
x_test
y_test
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
y_predict=model.predict([[22]])
y_predict
```

```python
x_pred = model.predict(x_train)
x_pred
```

```python
y_test[:6]
model.predict(x_test[:8])
```

```python
plt.scatter(x_train,y_train,color='blue')
plt.plot(x_train,x_pred,color='pink')
plt.title('Salary vs Experience (Training Dataset)')
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.show()
```

```python
plt.scatter(x_test,y_test,color='green')
plt.plot(x_train,x_pred,color='blue')
plt.title("Salary vs Experience (Training Dataset)")
plt.xlabel("YearsExperience")
plt.ylabel("Salary(in Ruppess)")
plt.show()
```
