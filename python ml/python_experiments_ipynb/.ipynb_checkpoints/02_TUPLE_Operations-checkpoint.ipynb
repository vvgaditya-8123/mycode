# Experiment 2 — TUPLE Operations
**Source:** tuple PDF

Various operations using Tuple data structure in Python. `Aditya` has been changed to `Aditya`.

```python
mytuple=('manish',)
print(mytuple)
```

```python
mytuple=('manish','gupta','amity','university')
print(type(mytuple))
print(mytuple)
```

```python
mytuple=('manish','gupta','amity','university')
mytuple1=list(mytuple)
mytuple1[1]='Aditya'
print(type(mytuple1))
mytuple=tuple(mytuple1)
print(mytuple)
print(type(mytuple))
```

```python
mytuple=('manish','gupta','amity','university')
print(mytuple[3])
print(mytuple[-3])
print(mytuple[1:3])
print(mytuple[-3:-1])
print(mytuple[1:])
```

```python
mytuple=('manish','gupta','amity','university')
mytuple1=list(mytuple)
mytuple1.insert(1,'Aditya')
mytuple1.append('Aditya')
mytuple=tuple(mytuple1)
print(mytuple)
```
