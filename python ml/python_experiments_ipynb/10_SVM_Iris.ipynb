# Experiment 10 — Support Vector Machine on Iris Dataset
**Source:** Support Vector Regression PDF

The PDF labels the experiment as Support Vector Machine and uses `SVC` for classification on the Iris Dataset.

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
```

```python
dir(iris)
```

```python
iris.target
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
df['target'] = iris.target
df.head()
```

```python
from matplotlib import pyplot as plt
%matplotlib inline

df0 = df[df.target==0]
df1 = df[df.target==1]
df2 = df[df.target==2]

print(df0.shape)
print(df1.shape)
print(df2.shape)
```

```python
df2.head()
```

```python
print(df0.columns)

import matplotlib.pyplot as plt
plt.xlabel('sepal length (cm)')
plt.ylabel('sepal width (cm)')
plt.scatter(df0['sepal length (cm)'], df0['sepal width (cm)'], marker='+', color='red')
plt.scatter(df1['sepal length (cm)'], df1['sepal width (cm)'], marker='o', color='green')
plt.scatter(df2['sepal length (cm)'], df2['sepal width (cm)'], marker='*', color='black')
plt.show()
```

```python
plt.xlabel('petal length (cm)')
plt.ylabel('petal width (cm)')
plt.scatter(df0['petal length (cm)'], df0['petal width (cm)'], marker='+', color='red')
plt.scatter(df1['petal length (cm)'], df1['petal width (cm)'], marker='o', color='green')
plt.scatter(df2['petal length (cm)'], df2['petal width (cm)'], marker='*', color='black')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

x = df.iloc[:,:-1]
y = df.target

x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.2,shuffle=False
)
```

```python
from sklearn.svm import SVC

# model = SVC()
# model = SVC(C=20)
# model = SVC(C=1000, gamma=5)

model = SVC(C=50, kernel='linear')
model.fit(x_train,y_train)
model.score(x_test,y_test)
```

```python
C_values = [1,2,5,10,20,30,50,80,100]
gamma_values = [1,2]
results = []

for C in C_values:
    for gamma in gamma_values:
        model = SVC(C=C, gamma=gamma)
        model.fit(x_train,y_train)
        accuracy = model.score(x_test,y_test)
        results.append([C,gamma,accuracy])
        print("C=",C,"gamma=",gamma,"Accuracy=",accuracy)

results
```
