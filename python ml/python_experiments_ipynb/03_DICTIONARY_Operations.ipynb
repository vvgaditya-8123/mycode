Various operations using Dictionary data structure in Python. `Aditya` has been changed to `Aditya`.

```python
dic={'manish':'teacher','Aditya':'student'}
print(dic)
print(dic['manish'])
dic['manish']
```

```python
dic={'manish':'teacher','Aditya':'student'}
dic.keys()
```

```python
dic={'manish':'teacher','Aditya':'student'}
print(dic.keys())
print(dic.items())
```

```python
dic={'manish':'teacher','Aditya':'student'}
dic.items()
```

```python
dic={'manish':'teacher','Aditya':'student'}
dic.values()
```

```python
dic={'manish':'teacher','Aditya':'student'}
del dic['manish']
print(dic)
dic.pop('Aditya')
print(dic)
```

```python
dic={'manish':'teacher','Aditya':'student'}
for x in dic:
    print(x)
```

```python
dic={'manish':'teacher','Aditya':'student'}
for x in dic:
    print(dic[x])
```

```python
dic={'manish':'teacher','Aditya':'student'}
for x in dic.values():
    print(x)
```

```python
dic={'manish':'teacher','Aditya':'student'}
for x in dic.items():
    print(x)
```

```python
dic={'manish':'teacher','Aditya':'student'}
emp={'v':'vyom'}
dic.update(emp)
print(dic)
```
